##  Additional Tasks

### 21. Using groupby, find the total quantity ordered per item_name. Show top 5 most ordered items.

```python
top_5 = chipo.groupby('item_name')['quantity'].sum().sort_values(ascending=False).head(5)
print(top_5)
```

### 22. How many orders contain more than 3 items? (group by order_id and sum quantity)

```python
order_counts = chipo.groupby('order_id')['quantity'].sum()
big_orders = (order_counts > 3).sum()
print(f"Orders with > 3 items: {big_orders}")
```

### 23. Create a 5x5 matrix of random integers between 1 and 100. Find the row with the highest sum.

```python
mat = np.random.randint(1, 101, (5, 5))
# Sum each row (axis=1) and find the index of the maximum sum
max_row_idx = np.argmax(mat.sum(axis=1))
print(f"Matrix:\n{mat}")
print(f"Highest sum is in row: {max_row_idx}")
```
